## Entendimento da ideia (síntese)

**Problema:** artistas e gravadoras não têm uma forma sistemática de saber, antes ou logo após o lançamento, se uma faixa tem as características estruturais associadas a hits — nem de identificar faixas já lançadas com potencial "escondido" (alta similaridade a hits, mas baixa popularidade atual).

**Dado disponível:** características de áudio do Spotify (`danceability`, `energy`, `valence`, `tempo`, `loudness`, `acousticness`, `instrumentalness`, `liveness`, `speechiness`) + `popularity` + `track_genre`, para ~114 mil faixas (dataset Kaggle, `archive/dataset.csv`).

**Solução proposta** (já esboçada no pipeline reaproveitado):
1. Definir "hit" operacionalmente — top X% de `popularity`, global ou por gênero (Frente 2/3).
2. Treinar um classificador (Random Forest) que estima a probabilidade de uma faixa ser hit a partir das features de áudio + gênero (Frente 3).
3. Extrair quais features mais pesam nessa decisão (`feature_importances_`) (Frente 3).
4. Complementar com clustering (KMeans) para descrever "arquétipos sonoros" e sua taxa de hits (Frente 3/4).
5. Empacotar tudo em funcionalidades de produto: comparar uma faixa com o "perfil de hit" do seu gênero, e listar faixas subestimadas — baixa popularidade, alta probabilidade/similaridade a hit (Frente 4).

**O que ainda não está decidido** (detalhado no módulo 02): threshold de "hit", estratégia de encoding de gênero, se o modelo roda pré ou pós-lançamento (implicação de vazamento de dados), formato final de entrega do produto, e toda a frente Ética.
